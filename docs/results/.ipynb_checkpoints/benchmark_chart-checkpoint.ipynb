# GGM Tree Benchmark Plots

This notebook loads the benchmark output and makes the three plots used in the report.

````python
from pathlib import Path
import sys

repo = Path.cwd()
while not (repo / "bench" / "plot_results.py").exists():
    if repo.parent == repo:
        raise FileNotFoundError("Could not find project folder")
    repo = repo.parent

sys.path.insert(0, str(repo / "bench"))

from plot_results import (
    load_results,
    plot_speedup_vs_depth,
    plot_throughput_comparison,
    plot_runtime_vs_depth,
    choose_throughput_depth,
)

benchmark_file = repo / "output" / "benchmark_ggm_output.txt"
if not benchmark_file.exists() or benchmark_file.stat().st_size == 0:
    benchmark_file = repo / "docs" / "benchmark_ggm_output.txt"
if not benchmark_file.exists() or benchmark_file.stat().st_size == 0:
    benchmark_file = repo / "output" / "benchmark_ggm_8_12_16_20.txt"
if not benchmark_file.exists() or benchmark_file.stat().st_size == 0:
    benchmark_file = repo / "output" / "benchmark_ggm.txt"

figure_dir = repo / "docs" / "results" / "figures"

rows = load_results(benchmark_file)
depths = sorted(set(row["depth"] for row in rows))

print("Loaded:", benchmark_file)
print("Depths in this file:", depths)
print("Figures will be saved to:", figure_dir)
````

## Plot 1: Speedup vs Depth

This plot compares GPU speedup over CPU for Spongent and Keccak on the same graph. The star marks the first depth where GPU is faster than CPU.

````python
path = plot_speedup_vs_depth(rows, figure_dir)
print("Saved:", path)
````

### Conclusion: Speedup vs Depth

GPU helps more when the tree is large enough. Keccak gets GPU speedup earlier, while Spongent needs a larger depth before GPU becomes faster.

## Plot 2: Throughput Comparison

This plot compares CPU-Spongent, GPU-Spongent, CPU-Keccak, and GPU-Keccak at one depth. The y-axis uses log scale because Keccak throughput is much larger.

````python
depth = choose_throughput_depth(rows)
path = plot_throughput_comparison(rows, figure_dir, depth)
print("Using depth:", depth)
print("Saved:", path)
````

### Conclusion: Throughput Comparison

Higher throughput means more GGM leaves are generated per second. In our benchmark, Keccak has much higher throughput than Spongent, especially on GPU.

## Plot 3: Runtime vs Depth

This plot shows runtime growth as tree depth increases. It has four lines: CPU-Spongent, GPU-Spongent, CPU-Keccak, and GPU-Keccak.

````python
path = plot_runtime_vs_depth(rows, figure_dir)
print("Saved:", path)
````

### Conclusion: Runtime vs Depth

Runtime grows quickly as depth increases because the number of leaves doubles every level. GPU runtime grows slower for the larger trees because many nodes can run in parallel.

## Reproduce

The same plots can be generated outside the notebook with:

```bash
python bench/plot_results.py output/benchmark_ggm_8_12_16_20.txt
```

If that file does not exist yet, the script will use `output/benchmark_ggm.txt`.
